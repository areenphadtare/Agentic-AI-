```python
from dotenv import load_dotenv
from pathlib import Path
import os

env_path = Path("/content/.env")
load_dotenv(env_path, override=True)

Google_API_Key = os.getenv("Google_API_KEY")

print("API key loaded:", bool(Google_API_Key))
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    google_api_key=Google_API_KEY,
    temperature=0
)

print("Gemini model initialized:", model is not None)
```

```python
#create a state
class LLMState(TypedDict):
    input_text: str
    output_text: str
```

```python
def llm_qa(state: LLMState) -> LLMState:
    # Extract the question from the state
    question = state["input_text"]

    # Form a prompt for the LLM
    prompt = f"Answer the following question: {question}"

    # Ask the LLM to answer the question
    response = model.invoke(prompt).content

    # Update the state with the answer
    state["output_text"] = response

    return state
```

```python
#Create a simple state graph for LLM workflow
graph = StateGraph(LLMState)

graph.add_node('llm_qa',llm_qa)

graph.add_edge(START,'llm_qa')
graph.add_edge('llm_qa',END)

workflow = graph.compile()
```

```python
initial_state = {
    "input_text": "How far is the Earth from the Sun?",
}

final_state = workflow.invoke(initial_state)

print(final_state)
```
