```python
from langgraph.graph import StateGraph ,START,END
from langchain_openai import ChatOpenAI
from typing import TypedDict, Any
from dotenv import load_dotenv
 # Load environment variables from .env file
```

```python
load_dotenv()
```

```python
import os 
api_key =os.getenv("Google_API_KEY")
```

```python
model=api_key
```

```python
#create a state
class LLMState(TypedDict):
    input_text: str
    output_text: str
```

```python
def llm_qa(state: LLMState) -> LLMState:
    # Extract the question from the state
    question = state["input_text"]

    # Form a prompt for the LLM
    prompt = f"Answer the following question: {question}"

    # Ask the LLM to answer the question
    response = model.invoke(prompt).content

    # Update the state with the answer
    state["output_text"] = response

    return state
```

```python
#Create a simple state graph for LLM workflow
graph = StateGraph(LLMState)

graph.add_node('llm_qa',llm_qa)

graph.add_edge(START,'llm_qa')
graph.add_edge('llm_qa',END)

workflow = graph.compile()
```

```python
initial_state = {
    "input_text": "How far is the Earth from the Sun?",
    "output_text": ""
}

final_state = workflow.invoke(initial_state)

print(final_state["output_text"])
```
